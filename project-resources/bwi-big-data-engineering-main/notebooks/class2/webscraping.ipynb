```python
import requests
```

```python
r = requests.get("https://www.immobilienscout24.at/regional/1020/wohnung-kaufen")
```

```python
from bs4 import BeautifulSoup
```

```python
soup = BeautifulSoup(r.text, 'html.parser')
```

```python
listings = soup.find_all('li', {'class': li_class})
```

```python
for listing in listings:
    price_info = listing.find('li', {'class': 'Text-color--gray-dark-sLZ'})
    print(price_info.text)
```
